## Operadores de Comparación
Los siguientes se pueden usar para comparar variables (ya sean cadenas de texto, flotantes, enteros, ...)
Comparan dos valores y retornan `True` o `False`.

| Operador | Descripción | Ejemplo |
|----------|-------------|--------|
| `==` | Igual a | `5 == 5` → `True` |
| `!=` | No igual a | `5 != 3` → `True` |
| `>` | Mayor que | `5 > 3` → `True` |
| `<` | Menor que | `5 < 3` → `False` |
| `>=` | Mayor o igual | `5 >= 5` → `True` |
| `<=` | Menor o igual | `5 <= 4` → `False` |
| `is` | Mismo objeto en memoria | |
| `in` | Contenido en secuencia | |

Al usarlos el resultado será un valor tipo bool (`True` , o `False`)

```python
numero_1 = 50
numero_2 = 51

print(numero_1 == numero_2)
print(numero_1 != numero_2)
```

# Ejercicio

- crear un arreglo con las horas de 0 a 24
- crear un arreglo para almacenar la temperatura
- utilizar un ciclo for para clcular la temperatura en cada hora
- graficar

$T(t) = 20 + 5sin(\frac{2 \pi t}{24})$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# crear el primer arreglo

t = np.linspace(0,24, 100)
T = np.zeros(100)

for i in range(0, len(t)):
  T[i] = 20 + 5*np.sin(2*np.pi*t[i]/24)
```

```python
plt.figure(figsize = (15,5))

plt.plot(t,T)


plt.show()
```
